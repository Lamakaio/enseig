```python
import pandas as pd
import geopandas as gpd
```

```python
gdf = gpd.read_file("GEO_TRA_BDTO_axes_ferroviaires_itineraires_normandie.dbf")
```

```python
gdf["LONGUEUR"] = gdf["geometry"].map(lambda x: x.length)
```

```python
gdf = gdf[["DEPART", "ARRIVEE", "LONGUEUR"]]
```

```python
from sqlalchemy import create_engine

engine = create_engine('sqlite:///TrainNormandie.db', echo=False)
```

```python
gdf.to_sql(name='Lignes', con=engine)
```

```python
from sqlalchemy import text

with engine.connect() as conn:
   print(conn.execute(text("SELECT * FROM Lignes")).fetchall())
```
