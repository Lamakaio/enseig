## Le jeu de la vie

Le jeu de la vie est ce qu'on appelle un _automate cellulaire_. C'est un ensemble de cellules (= des cases) sur une grille infinie, qui évoluent selon des règles simples : 
- les cellules peuvent être vivante (noire et `True` ici) ou morte (blanche et `False` ici)
- si une cellule morte a _exactement_ 3 voisines vivantes, elle deviens vivante.
- si une cellule vivante n'a pas 2 ou 3 cellules voisines vivante, elle meurt.

On va ici implémenter les règles du jeu de la vie. 

Au lieu de prendre une grille infinie, on va prendre une grille de 30 par 30, et on va supposer qu'elle "boucle" : à droite de la dernière cellule d'une ligne, on a la première cellule de cette même ligne.
