## I. Généralitées

On étudiera dans ce TP des arbres contenant des _chaines de caractère_, on `string` en OCaml. 

!!! question Question 1
Définissez un type `arbre` permettant de représenter un arbre binaire stockant des chaines de caractères dans ses noeuds et feuilles.

Puis, avant de continuer, regardez la case réponse et utilisez le type donné (pour éviter les problèmes).

??? info Réponse
```ocaml
type arbre = 
    | F of string
    | N of arbre * string * arbre (*l'ordre exact pourrait être différent, mais pour ce TP on utilisera celui-là*)

```

!!! question Question 8 
Ecrivez une fonction `ppac: arbre -> string -> string -> arbre` (pour plus proche ancêtre commun), qui prend en argument un arbre, et deux chaines de caractères qu'on suppose présentes dans l'arbre, et renvoie l'arbre dont la racine est l'ancêtre commun le plus proche entre les deux (c'est à dire le plus petit sous-arbre qui contient les deux espèces).

On ne cherchera pas à faire une fonction optimisée.
!!!

??? info indications
- Un noeud est le plus proche ancêtre commun entre "a" et "b", si "a" est dans sont fils gauche et "b" dans son fils droit, ou vice-versa.
- On peut utiliser la fonction `recherche` pour chercher "a" dans le fils droit. 
- Si "a" et "b" sont dans le même fils, c'est que l'ancêtre commun est aussi dans ce fils.
